```python
# IRCTC.NS fundamental & 6M forecast using yfinance
# -------------------------------------------------
# Prereqs: pip install yfinance pandas numpy
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

def run_analysis(TICKER):
    today = datetime.today().date()

    tkr = yf.Ticker(TICKER)

    # --- Fundamentals snapshot ---
    info = tkr.info  # dict with metadata; keys can vary
    price = info.get("currentPrice", np.nan)
    market_cap = info.get("marketCap", np.nan)
    pe_ttm = info.get("trailingPE", np.nan)
    eps_ttm = info.get("trailingEps", np.nan)
    dividend_rate = info.get("dividendRate", 0.0)
    dividend_yield = info.get("dividendYield", 0.0)  # fraction (e.g., 0.0128 for 1.28%)

    # --- Price history & vol ---
    hist = tkr.history(period="5y", interval="1d")
    hist = hist.dropna()
    # log returns
    ret = np.log(hist["Close"]).diff().dropna()
    mu = ret.mean() * 252           # annualised drift (naive)
    sigma = ret.std() * 252**0.5    # annualised vol from daily data
    last_close = hist["Close"].iloc[-1]

    # --- Monte Carlo (GBM) ---
    np.random.seed(42)
    days = 126  # ~6 months trading days
    n_paths = 10000
    dt = 1/252
    drift = (mu - 0.5 * sigma**2) * dt
    shock_scale = sigma * np.sqrt(dt)

    Z = np.random.normal(0, 1, (days, n_paths))
    paths = np.zeros((days+1, n_paths))
    paths[0] = last_close
    for t in range(1, days+1):
        paths[t] = paths[t-1] * np.exp(drift + shock_scale * Z[t-1])

    final_prices = paths[-1]
    median_fp = float(np.median(final_prices))
    p05 = float(np.percentile(final_prices, 5))
    p95 = float(np.percentile(final_prices, 95))

    # --- Rule-based headline sentiment (lightweight) ---
    news = tkr.news if hasattr(tkr, "news") else []
    pos_words = {"growth","profit","up","rise","dividend","record","approval","strong","steady"}
    neg_words = {"down","decline","loss","fall","slip","weak","sell","stake sale","OFS","volatility"}


    def headline_score(h):
        text = (h.get("title","") + " " + h.get("summary","")).lower()
        score = sum(w in text for w in pos_words) - sum(w in text for w in neg_words)
        return score

    sent_scores = [headline_score(n) for n in news[:50]]
    sentiment = np.mean(sent_scores) if sent_scores else 0.0  # >0 bullish, <0 bearish

    # --- Transparent hold decision (0/1) ---
    # Criteria (tweak as you like):
    # 1) Valuation: P/E not extremely stretched (>45 considered stretched here)
    # 2) Forecast median >= last price (positive drift)
    # 3) Sentiment non-negative
    hold = int((pe_ttm and pe_ttm < 45) and (median_fp >= last_close) and (sentiment >= 0))

    summary = {
        "as_of": str(today),
        "ticker": TICKER,
        "last_close": round(last_close, 2),
        "market_cap": market_cap,
        "pe_ttm": pe_ttm,
        "eps_ttm": eps_ttm,
        "dividend_rate": dividend_rate,
        "dividend_yield": round(dividend_yield * 100, 2) if dividend_yield else None,
        "annualised_mu": round(mu, 4),
        "annualised_sigma": round(sigma, 4),
        "forecast_median_6m": round(median_fp, 2),
        "forecast_p05_6m": round(p05, 2),
        "forecast_p95_6m": round(p95, 2),
        "sentiment_score": round(sentiment, 3),
        "HOLD_boolean": hold
    }

    print(summary)
    return hold



for index, row in df1.iterrows():
    print(f"Row {index}:")
    run_analysis(row["tradingsymbol"] + ".NS")


```
