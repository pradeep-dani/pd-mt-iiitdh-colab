```python
# Problem 1 — Constructing a Grouped Frequency Distribution
# Question:
# The following are the marks (out of 50) obtained by 20 students in a test:
# 12, 25, 34, 45, 22, 17, 30, 48, 29, 41,
# 33, 26, 19, 28, 39, 44, 31, 24, 15, 37
# Construct a grouped frequency distribution table using class intervals of width 10,
# starting from 10–20.


# List of marks obtained by 20 students
marks = [12, 25, 34, 45, 22, 17, 30, 48, 29, 41,
         33, 26, 19, 28, 39, 44, 31, 24, 15, 37]


# Define the class intervals starting from 10–20 with width 10
class_intervals = [(i, i + 10) for i in range(10, 51, 10)]
print("class_intervals", class_intervals);
print("\n");

# Initialize a dictionary to store frequency distribution
frequency_distribution = {f"{start}-{end}": 0 for start, end in class_intervals}
print("frequency_distribution", frequency_distribution);
print("\n");

# Count the frequency of marks in each class interval
for mark in marks:
    for start, end in class_intervals:
        if start <= mark < end:
            frequency_distribution[f"{start}-{end}"] += 1
            break


# Print the grouped frequency distribution table
print("Grouped Frequency Distribution Table:")
print("{:<15} {:<10}".format("Class Interval", "Frequency"))
for interval, frequency in frequency_distribution.items():
    print("{:<15} {:<10}".format(interval, frequency))
```

```python
# Problem 2 — Relative Frequency Distribution
# Question:
# Using the above frequency table, calculate the relative frequency for each class


# Calculate total number of observations
total_observations = sum(frequency_distribution.values())
print("total_observations", total_observations);
print("\n");


# Calculate relative frequencies
relative_frequencies = {interval: freq / total_observations for interval, freq in frequency_distribution.items()}
print("relative_frequencies", relative_frequencies);
print("\n");
```

```python
# Problem 3 — Cumulative Frequency Distribution (Less-Than Type)
# Question:
# Construct the less-than cumulative frequency for the dataset in Problem 1.


# Calculate less-than cumulative frequencies
cumulative_frequencies = []
cumulative_sum = 0

for interval, frequency in frequency_distribution.items():
    cumulative_sum += frequency
    cumulative_frequencies.append(cumulative_sum)
print("cumulative_sum", cumulative_sum);
print("\n");

print("cumulative_frequencies", cumulative_frequencies);
print("\n");

# Combine the results into a list of tuples
less_than_cumulative_frequency = list(zip(class_intervals, cumulative_frequencies))

print("less_than_cumulative_frequency", less_than_cumulative_frequency);
print("\n");

# Display the less-than cumulative frequency distribution
for interval, cum_freq in less_than_cumulative_frequency:
    print(f"Less than {interval}: {cum_freq}")

```

```python
# Problem 4 — Coefficient of Variation
# Question:
# A dataset has a mean (X̄) = 40 and a standard deviation (σ) = 8.
# Find the Coefficient of Variation (CoV) and interpret the result.


# Given values
mean = 40
std_dev = 8

# Calculate Coefficient of Variation (CoV)
# Formula: CoV = (Standard Deviation / Mean) * 100%
if mean != 0:
    cov = (std_dev / mean) * 100
    print(f"Mean (X̄): {mean}")
    print(f"Standard Deviation (σ): {std_dev}")
    print(f"Coefficient of Variation (CoV): {cov:.2f}%")
else:
    print("Cannot calculate CoV: Mean is zero.")
```

```python
# Problem 5: Correlation Coefficient
# Question:
# The following table shows the marks obtained by 5 students in Mathematics and Physics:

# Student Math (X) Physics (Y)
# A 35 30
# B 40 40
# C 45 50
# D 50 45
# E 55 60

# Calculate the Karl Pearson’s correlation coefficient (r) between marks in Mathematics and Physics.

import numpy as np

# Marks obtained by students in Mathematics and Physics
math_marks = [35, 40, 45, 50, 55]
physics_marks = [30, 40, 50, 45, 60]

print("math_marks", math_marks);
print("\n");
print("physics_marks", physics_marks);
print("\n");

# Calculate Karl Pearson’s correlation coefficient
correlation_matrix = np.corrcoef(math_marks, physics_marks)
correlation_coefficient = correlation_matrix[0, 1]

# Print the result
print(f"Karl Pearson’s correlation coefficient (r) between marks in Mathematics and Physics is: {correlation_coefficient:.4f}")
```

```python
# Problem 6 — Histogram Construction
# Question:
# The table below shows the number of students who scored marks (out of 100) in a Data
# Science exam.
# Marks Range Frequency
# 0–20 4
# 20–40 6
# 40–60 10
# 60–80 8
# 80–100 2
# Construct a Histogram representing the frequency distribution.
# Identify the class interval with the highest frequency (the mode class).


import matplotlib.pyplot as plt

# Data from the frequency distribution
marks_range = ['0–20', '20–40', '40–60', '60–80', '80–100']
frequencies = [4, 6, 10, 8, 2]

print("marks_range", marks_range);
print("\n");
print("frequencies", frequencies);
print("\n");

# Create the histogram using bar plot
plt.figure(figsize=(8, 6))
plt.bar(marks_range, frequencies, color='skyblue', edgecolor='black')
plt.title('Histogram of Marks in Data Science Exam')
plt.xlabel('Marks Range')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Save the histogram as an image
plt.savefig('data_science_histogram.png')

# Identify the mode class
max_freq = max(frequencies)
mode_class = marks_range[frequencies.index(max_freq)]
print(f"The mode class (highest frequency) is: {mode_class} with {max_freq} students.")
```

```python
# Problem 7 — Bar Plot Comparison
# Question:
# The table below shows the number of publications made by three departments over four
# years.
# Year CS Dept ECE Dept Mech Dept
# 2022 25 20 15
# 2023 30 25 20
# 2024 35 30 25
# 2025 40 28 30
# Plot a grouped bar chart comparing the departments over the years.
# Which department shows the highest overall growth?

import pandas as pd
import matplotlib.pyplot as plt

# Create the dataset
data = {
    'Year': [2022, 2023, 2024, 2025],
    'CS Dept': [25, 30, 35, 40],
    'ECE Dept': [20, 25, 30, 28],
    'Mech Dept': [15, 20, 25, 30]
}

# Convert to DataFrame
df = pd.DataFrame(data)

print("Data", df);
print("\n");

# Set the figure size
plt.figure(figsize=(10, 6))

# Set the positions and width for the bars
bar_width = 0.25
r1 = range(len(df['Year']))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]

# Plot the bars
plt.bar(r1, df['CS Dept'], color='blue', width=bar_width, label='CS Dept')
plt.bar(r2, df['ECE Dept'], color='green', width=bar_width, label='ECE Dept')
plt.bar(r3, df['Mech Dept'], color='red', width=bar_width, label='Mech Dept')

# Add labels and title
plt.xlabel('Year')
plt.ylabel('Number of Publications')
plt.title('Publications by Department Over Years')
plt.xticks([r + bar_width for r in range(len(df['Year']))], df['Year'])
plt.legend()

# Save the plot
plt.tight_layout()
#plt.savefig('department_publications_comparison.png')
plt.show()

# Determine department with highest overall growth
growth = {
    'CS Dept': df['CS Dept'].iloc[-1] - df['CS Dept'].iloc[0],
    'ECE Dept': df['ECE Dept'].iloc[-1] - df['ECE Dept'].iloc[0],
    'Mech Dept': df['Mech Dept'].iloc[-1] - df['Mech Dept'].iloc[0]
}

highest_growth_dept = max(growth, key=growth.get)
print(f"The department with the highest overall growth is: {highest_growth_dept}")
```
